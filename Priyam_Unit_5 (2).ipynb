```python
#w.a.p. to perform a circular shift on a list to the right direction 
#shift is to be entered by user

l = [1,2,3,4,5,6,7]
shift = int(input("enter shift:"))
while shift>0:
    x = l.pop()
    l.insert(0,x)
    shift -= 1
print(l)

# OR
shift = shift%len(l)
x = l[-shift:]+l[:-shift]
print(x)
```

```python
l = [5,5,2,5,8]

count = 0
for i in range(len(l)):
    y = l.copy()
    y.pop(i)
    even = sum(y[::2])
    odd = sum(y[1::2])
    if even==odd:
        count+=1
        print("Index to remove:",i)
        print("list after removing that index:",l[:i]+l[i+1:])
print("Total special elements:",count)
```

```python
#w.a.p. that read the word from the paragraph and store them as keys in dictionary 
# and count the frequency of it as a value
s = "dog the quick brown for jumps over the lazy dog"
words = (s.split())
d = {}
for word in words:
    if word in d:
        d[word] +=1
    else:
        d[word] = 1

print(d)
```

```python
# w.a.p. to create a dictionary with key as first character of the word and value as a list of words starting with 
# that character 

s = "dog the quick brown for jumps over the lazy dog"
d = {}
words = s.split()
for i in words:
    key = i[0]
    if key in d:
        d[key].append(i)
    else:
        d[key] = [i] 
print(d)
```

```python
# w.a.p. to replace the word of a given string from the dictionary 
s = 'CampusX best for DS students'
d = {'best':'is the best channel','DS':'Data Science'}
ans = ""
words = s.split()
for word in words:
    if word in d:
        ans += d[word] + ' '
    else:
        ans += word + ' '
print(ans)
```

# SET


- mutable
- if we want to represent a group of unique value as a single entity then we should go for set
- duplicates not allowed
- insertion order not preserved but we can sort the elements
- indexing and slicing not allowed for set
- heterogeneous elements allowed
- we can represent set elements within curly brackets with comma separation
- we can apply mathematical operations like union , interesection , difference etc. on set

```python
# creation of set 

s = {10,10,20,30,40,30}
print(s)
print(type(s))
```

```python
l = [1,2,3,4,4,5,5,6,7]
s = set(l)
print(s)
```

```python
s = set(range(5))
print(s)
```

```python
s = set()
print(s)
```

```python
s = eval(input("Enter set:"))
print(s)
```

```python
# mathematical operations of set

#  union()
x = {10,20,30,40}
y = {30,40,50,60}
print(x.union(y))
print(x|y)
print()

# intersection()
x = {10,20,30,40}
y = {30,40,50,60}
print(x.intersection(y))
print(x & y)
print()

# difference()
x = {10,20,30,40}
y = {30,40,50,60}
print(x.difference(y))
print(x-y)
print()

# symmetric_difference()
x = {10,20,30,40}
y = {30,40,50,60}
print(x.symmetric_difference(y))
print(x^y)
print()

# issubset()
x = {"a","b","c"}
y = {"d","e","a","h","b","c"}
print(x.issubset(y))
print(y.issubset(x))
print()

#issuperset()
x = {"a","b","c"}
y = {"d","e","a","h","b","c"}
print(x.issuperset(y))
print(y.issuperset(x))
print()

# membership operator

s = set("LJ University")
print(s)
print("l" in s)
print("x" not in s)
print("U" in s)
print()
```
