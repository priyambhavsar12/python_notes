### Unit - 5:

#### Mutable Data Structure


```python
#list

l = [1,2,3,4]
l[2] = 10
print(l)
```

```python
# duplicates values are allowed 
# insertion order preserved
# list is dynamic
```

```python
#creation of list

l = []
print(l)
l = [1]
print(l) # single value list
l = [1,2,]
print(l)
l = list(range(1,10))
print(l)
s = "Yo , Im Priyam"
l = list(s)
print(l)
```

```python
# Dynamic input          eval func-> for dynamic input
l = eval(input("Enter Data:"))
print(l)
print(type(l))
```

```python
# accessing elements in list

# 1. by using index
l = [10,20,30,40,50]
print(l[1])
print(l[-1])

# 2. by using slicing operator

n = list(range(1,11))
print(n[2:7:2])
print(n[4::2])
print(n[3:7])
print(n[8:2:-2])
print(n[4:100])
print(n[::-1])
```

```python
# aliasing and cloning of list

l = [1,2,3,4,5]
y = l.copy()
y[2] = 30
print(l)
print(id(y))
print(id(l))

```

```python
l1 = [1,2,3,4,5]
l2 = [10,20]

print(l1+l2)
print(l1*3)
```

```python
#important functions of list

#len()
l = [1,2,3,4]
print(len(l))
```

```python
# count()
l = [1,1,2,2,2,3,3]
print(l.count(2))
print(l.count(4))
```

```python
# append()
l = ["A","B","C"]
l.append("D")
l.append("E")
print(l)
```

```python
# insert()
n = [1,2,3,4,5]
n.insert(-1,30) # index,element
print(n)
```

```python
# extend()
l1 = [10,20,30]
l2 = [40,50,60]
l1.extend(l2)
print(l1)
print(l2)
l2.extend(l1)
print(l1)
print(l2)
```

```python
#remove()
n = [1,2,1,3,2]
n.remove(1)
print(n)
n.remove(1)
print(n)
```

```python
# pop()
n = [10,20,10,40,50]
print(n.pop())
print(n.pop())
print(n.pop())
print(n.pop())
print(n.pop())
```

```python
n = [10,20,10,40,50]
n.pop(1)
n.pop()
print(n)
```

```python
#clear()
n = [10,20,10,40,50]
n.clear()
print(n)
```

```python
#reverse()
n = [1,2,3,4]
n.reverse()
print(n)
```

```python
#sort()
n = [10,20,10,40,50]
n.sort()
print(n)
n.sort(reverse = True)
print(n)

n = ["A","D","B","C"]
n.sort()
print(n)
```

```python
s = ["apple","cherry","strawberry","kiwi","banana"]
s.sort(key=len)
print(s)
```
