# ---------Working With Files--------

#### Files With Modes

- Read only(r)

- Read and write(r+)

- write only(w)

- write and read(w+)

- append only(a)

- append and read(a+)



##### read only(r)

- open text file for reading
- handle is positioned at beginning of the file
- if file does not exists , raises I/O error
- this is default access mode in which file is opened


```python
f = open('pfile1.txt','r')
print(f.read())
print(f.closed)
f.close()
print(f.closed)
```

```python
with open('pfile1.txt','r') as f:
    print(f.read())
    print(f.closed)
print(f.closed)
```

```python
with open('utf8_demo.txt','r',encoding='utf-8') as f:
    print(f.read())
```

```python
with open('pfile1.txt','r') as f:
    print(f.read())
```

##### read and write(r+)


```python
with open('pfile1.txt','r+') as f:
    f.write('yoooooo')
    print(f.read())
    f.write('LJ')
    print(f.read())
```

```python
f = open('pfile1.txt','r+')
f.write('heheh')
f.close()
```

```python
f = open('pfile1.txt','r+')
print(f.tell())
f.write("HUHHHHH")
print(f.tell())
f.seek(0)
print(f.tell())
f.close()
```

```python
f = open('pfile1.txt','r+')
f.read()
f.write('\nLJIET')
f.seek(0)
f.write("LJU")
print(f.read())
f.close()
```

```python
f = open('pfile1.txt','w')
f.write('LJU')
f.close()
```

```python
f = open('demo1.txt','w+')
f.write('LJ University')
f.write('\nLJU')
f.seek(0)
print(f.read())
f.close()
```

```python
f = open('demo1.txt','a')
f.write('\nNew Data In Append Mode')
f.seek(0)
f.write('\nNew Data after Seek to zero Index')
f.close()
```

### Reading the Files

- read() - returns the read bytes in the form of string
- readLine() - reads a line of the file and returns in form of string
- readlines() - reads all the lines and returns them as string element in list


```python
f = open('demo.txt','r')
print(f.read(3))
print(f.tell())
print(f.read(4))
print(f.tell())
print(f.read(5))
print(f.tell())
print(f.read())
print(f.tell())
f.close()
```

```python
f = open('demo.txt','r')
print(f.readline())
print(f.readline())
print(f.readline())
f.close()
```

```python
f = open('demo.txt','r')
print(f.readlines())
f.close()
```
