```python
#w.a.p. to sort the character of the string and first alphabet symbol followed by digit
s = "B4A1D3"
al = ""
digit = ""
for i in s:
    if i.isalpha():
        al += i
    elif i.isdigit():
        digit += i
x = sorted(al)
y = sorted(digit)
print(''.join(x+y))
```

```python
# s = "a4b3c2 " -> output -> "aaaabbbcc"

s = "a4b3c2"
al = ""
ans = ""
for i in s:
    if i.isalpha():
        al = i
    elif i.isdigit():
        ans = ans + al*int(i)
x = sorted(ans)
print(''.join(x))
```

```python
#w.a.p. to check the password entered by user is valid or not based on following conditions:
# 1. length must be 8 characters
# 2. alphabets must be between(a-z)
# 3.atleast one uppercase
# 4. atleast one digit
# 5. atleast one special character from (_,$,@)

l,u,sp,d=0,0,0,0
password = input("enter password:")
if len(password)>=8:
   for i in password:
        if i.isupper():
           u+=1
        if i.islower():
           l+=1
        if i.isdigit():
           d+=1
        if i=='$' or i=='@' or i=='_':
           sp+=1
   if l>0 and u>0 and d>0 and sp>0 and (u+l+d+sp)==len(password):
        print("Valid Password")
   else:
        print("Invalid Password")

       
else:
    print("Invalid Password,Password must be atleast 8 characters")
```

##### Tuple 


- immutable

- read only version of list

- insertion order is preserved

- duplicates are allowed

- heterogeneous objects are allowed

- differentiate duplicate objects by using index

- +ve and -ve index supported

- represent tuple elements within paranthesis (()) with comma separation

- parenthesis are optional but recommended to use

```python
t = ((1,2,3),(4,5,6),1,"yo",True)
print(t)
```

```python
# Creation of Tuple

t = (1,2,3,4,5)
print(type(t))
```

```python
t = ()
print(type(t))
```
