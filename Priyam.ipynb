```python
# Changing the case of String

# 1. lower()

s = "PRIYAM"
t = s.lower()
print(t)

# 2. upper()
s = "priyam"
t = s.upper()
print(t)

# 3. swapcase()
s= "PriyAm"
t = s.swapcase()
print(t)

# 4. title()
s = "Python is very easy"
t = s.title()
print(t)

# 5. capitlize()

s = "Python Is Very Easy"
t = s.capitalize()
print(t)
```

```python
# sum of the digits and their average

s = "P@#$yt372h@$2735on"
sum = 0
count = 0
for i in range(len(s)):
    if s[i].isdigit():
        sum = sum+int(s[i])
        count+=1
        average = sum/count
print("sum is :",sum)
print("average :",average)
```

```python
# find() -> returns first occurence of that given substring, if not available we'll get -1

s = "123445678423"
s.find("4")

s = "Learning Python is very easy"
s.find("Python")
print(s.find("Java"))
```

```python
# index() -> returns first occurence as find() function but when not available throws error
s = "123445678423"
s.index("4")

s = "Learning Python is very easy"
s.index("Python")
#print(s.index("Java"))
print(s.index("e",21,))  # till end of string
```

```python
# count() -> counts the number of times substring is in a particular string

s = "abcdefjanafaenfanfaqnkfef"
print(s.count("a"))
print(s.count('f',5,10)) #between 5th to 10 th index
print(s.count('a',1,)) # till end of string
```

```python
# translate() with maketrans function
# -> removing punctuation using translate()

import string
print(string.punctuation)
print(len(string.punctuation))

s = "Py$@th#on"
x = s.maketrans("","",'$#')
print(x)
len(x)

print(s.translate(x))

# -> using a mapping table to replace characters

s = "Hello Sam!!"
table = s.maketrans("S","P",'!')
print(table)
print(s.translate(table))
s = "Hello Sam!"
table = s.maketrans("mSa","eJo")
print(table)
print(s.translate(table))
```

```python
#w.a.p. check the length of a string  if the length is even then print the first and last letter and if length is odd then first
#middle and last character
s  = str(input("enter string:"))
length = len(s)
if length%2==0:
    print(s[0]+s[length-1])
else:
    print(s[0]+s[length//2]+s[length-1])
```

```python
#w.a.p. to replace each special symbol with # to given string
import string
s= "/*Jon is @developer & musician!!"
for i in s:
    if i in string.punctuation:
        s = s.replace(i,"#")
print(s)
```

```python
#w.a.p. to replace each special symbol with # to given string with maketrans
s= "/*Jon is @developer & musician!!"
x = s.maketrans(string.punctuation,"#"*32)
print(s.translate(x))
```
